```python
system_prompt = f"""
# Your role

You are the digital twin of {name}, running on a website and chatting with visitors.
You represent {name} and answer questions about their career, background, skills and experience.

Here is a summary written by {name}:

{summary}

If asked, explain clearly that you are an AI, the digital twin of {name}.

# Context

Here is the text of {name}'s LinkedIn profile, so that you can answer questions:

{linkedin}

# Rules

Be professional and engaging, as if talking to a potential employer or client who found the website.
Do not answer questions unrelated to {name}'s career, background, skills and experience.
Steer the conversation back to professional topics.

Always stay in character as the digital twin of {name}.

IMPORTANT: If you don't know the answer, say so. Never make up an answer.
If the user asks about something not in the context above, say that you don't know.
"""
```

```python
def record_email(email):
    print(f"Tool called to record an email: {email}")
    with open("emails.txt", "a", encoding="utf-8") as f:
        f.write(email + "\n")
    return "Email received"
```

## describe the tool to the model as JSON


```python
record_email_json = {
    "name": "record_email",
    "description": "Use this tool to record that a visitor provided their email address",
    "parameters": {
        "type": "object",
        "properties": {
            "email": {
                "type": "string",
                "description": "The email address of this visitor",
            }
        },
        "required": ["email"],
        "additionalProperties": False,
    },
}
```

```python
tools = [{"type": "function", "function": record_email_json}]
```

```python
tools
```

```python
def chat(message, history):
    messages = [{"role": "system", "content": system_prompt}] + history + [{"role": "user", "content": message}]

    response = client.chat.completions.create(
        model="gpt-5.4-mini",
        messages=messages,
        tools=tools,
    )

    if response.choices[0].finish_reason == "tool_calls":
        assistant_message = response.choices[0].message
        tool_call = assistant_message.tool_calls[0]

        arguments = json.loads(tool_call.function.arguments)
        email = arguments.get("email")

        result = record_email(email)

        messages.append(assistant_message)
        messages.append({"role": "tool", "content": result, "tool_call_id": tool_call.id})

        response = client.chat.completions.create(
            model="gpt-5.4-mini",
            messages=messages,
            tools=tools,
        )

    return response.choices[0].message.content
```

```python
gr.ChatInterface(chat).launch(inbrowser=True)
```
